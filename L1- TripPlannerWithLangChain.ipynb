##### In this tutorial, we are building a simple trip planner using LangChain. The app takes a destination and a few preferences, then outputs a tailored trip plan. We'll use the LangChain Expression Language (LCEL) to chain together components such as prompt templates and LLMs.

```python
# importing the necessary libraries
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

from dotenv import load_dotenv
_ = load_dotenv()
```

```python
# use GPT-4 to process user input and generate trip plans.
model = ChatOpenAI(model="gpt-4o")
```

```python
# use a prompt template to set the context for the LLM to generate personalized trip plans based on user input.

system_template = """
You are a trip planner expert. Help me plan a trip to {destination}.
Consider my preferences for {preferences}.
"""

prompt_template = ChatPromptTemplate.from_messages([
    ('system', system_template),
    ('user', 'What should I do in {destination}?')
])
```

```python
# use a simple output parser to extract the LLM's response as a string.
parser = StrOutputParser()
```

```python
# now chain the prompt template, model, and output parser to process input and output
trip_planner_chain = prompt_template | model | parser
```

```python
# define a function to run a query through the trip planner
def plan_trip(destination, preferences):
    input_data = {"destination": destination, "preferences": preferences}
    return trip_planner_chain.invoke(input_data)
```

```python
# helper function to render markdown
from IPython.display import Markdown, display

def render_markdown(md_string):
    display(Markdown(md_string))
```

```python
# run the trip planner with different inputs
result = plan_trip("Paris", "museums, cafes, historical sites")
render_markdown(result)
```

```python
result = plan_trip("Tokyo", "technology, culture, nightlife")
render_markdown(result)
```
